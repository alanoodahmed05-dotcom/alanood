```python
# S3 - Classification Models
# Create a categorical target variable

data['Math_Result'] = data['math score'].apply(
    lambda score: 'Pass' if score >= 50 else 'Fail'
)

print(data[['math score', 'Math_Result']].head())
print(data['Math_Result'].value_counts())
```

```python
# Independent and Dependent Variables

X_class = data[['reading score', 'writing score']].values
y_class = data['Math_Result'].values

print("Independent Variables:")
print(X_class[:5])

print("Dependent Variable:")
print(y_class[:5])
```

```python
# Splitting the Dataset into Training and Testing

from sklearn.model_selection import train_test_split

X_class = data[['reading score', 'writing score']].to_numpy()
y_class = data['Math_Result'].to_numpy()

X_train_class, X_test_class, y_train_class, y_test_class = train_test_split(
    X_class,
    y_class,
    test_size=0.2,
    random_state=0,
    stratify=y_class
)

print("Training Set:", len(X_train_class))
print("Testing Set:", len(X_test_class))
```

```python
# Feature Scaling

from sklearn.preprocessing import StandardScaler

sc_class = StandardScaler()

X_train_scaled = sc_class.fit_transform(X_train_class)
X_test_scaled = sc_class.transform(X_test_class)

print(X_train_scaled[:5])
```

```python
# Model 1 - Logistic Regression

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Train the model
log_model = LogisticRegression(random_state=0)
log_model.fit(X_train_scaled, y_train_class)

# Predict the test data
y_pred_log = log_model.predict(X_test_scaled)

# Evaluate the model
print("Accuracy:", accuracy_score(y_test_class, y_pred_log))
print("Precision:", precision_score(y_test_class, y_pred_log, pos_label='Fail'))
print("Recall:", recall_score(y_test_class, y_pred_log, pos_label='Fail'))
print("F1-score:", f1_score(y_test_class, y_pred_log, pos_label='Fail'))
```

```python
# Model 2 - K-Nearest Neighbors (KNN)

from sklearn.neighbors import KNeighborsClassifier

knn_model = KNeighborsClassifier(n_neighbors=5)

# Train the model
knn_model.fit(X_train_scaled, y_train_class)

# Predict the test data
y_pred_knn = knn_model.predict(X_test_scaled)

# Evaluate the model
print("Accuracy:", accuracy_score(y_test_class, y_pred_knn))
print("Precision:", precision_score(y_test_class, y_pred_knn, pos_label='Fail', zero_division=0))
print("Recall:", recall_score(y_test_class, y_pred_knn, pos_label='Fail', zero_division=0))
print("F1-score:", f1_score(y_test_class, y_pred_knn, pos_label='Fail', zero_division=0))
```

```python
# Model 3 - Support Vector Machine (SVM)

from sklearn.svm import SVC

# Train the model
svm_model = SVC(kernel='rbf', random_state=0)
svm_model.fit(X_train_scaled, y_train_class)

# Predict the test data
y_pred_svm = svm_model.predict(X_test_scaled)

# Evaluate the model
print("Accuracy:", accuracy_score(y_test_class, y_pred_svm))
print("Precision:", precision_score(y_test_class, y_pred_svm, pos_label='Fail', zero_division=0))
print("Recall:", recall_score(y_test_class, y_pred_svm, pos_label='Fail', zero_division=0))
print("F1-score:", f1_score(y_test_class, y_pred_svm, pos_label='Fail', zero_division=0))
```

```python
# Model 4 - Decision Tree

from sklearn.tree import DecisionTreeClassifier

# Train the model
dt_model = DecisionTreeClassifier(random_state=0)
dt_model.fit(X_train_scaled, y_train_class)

# Predict the test data
y_pred_dt = dt_model.predict(X_test_scaled)

# Evaluate the model
print("Accuracy:", accuracy_score(y_test_class, y_pred_dt))
print("Precision:", precision_score(y_test_class, y_pred_dt, pos_label='Fail', zero_division=0))
print("Recall:", recall_score(y_test_class, y_pred_dt, pos_label='Fail', zero_division=0))
print("F1-score:", f1_score(y_test_class, y_pred_dt, pos_label='Fail', zero_division=0))
```

```python
# Model 5 - Random Forest

from sklearn.ensemble import RandomForestClassifier

# Train the model
rf_model = RandomForestClassifier(n_estimators=100, random_state=0)
rf_model.fit(X_train_scaled, y_train_class)

# Predict the test data
y_pred_rf = rf_model.predict(X_test_scaled)

# Evaluate the model
print("Accuracy:", accuracy_score(y_test_class, y_pred_rf))
print("Precision:", precision_score(y_test_class, y_pred_rf, pos_label='Fail', zero_division=0))
print("Recall:", recall_score(y_test_class, y_pred_rf, pos_label='Fail', zero_division=0))
print("F1-score:", f1_score(y_test_class, y_pred_rf, pos_label='Fail', zero_division=0))
```

```python
# S1 - Cluster Analysis
# Step 1: Select Features

X_cluster = data[['math score', 'reading score', 'writing score']]

print(X_cluster.head())
```

```python
# Step 2: Feature Scaling

from sklearn.preprocessing import StandardScaler

sc_cluster = StandardScaler()
X_cluster_scaled = sc_cluster.fit_transform(X_cluster)

print(X_cluster_scaled[:5])
```

```python
# Step 3: Elbow Method

from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

wcss = []

for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, random_state=0, n_init=10)
    kmeans.fit(X_cluster_scaled)
    wcss.append(kmeans.inertia_)

plt.plot(range(1, 11), wcss, marker='o')
plt.title('Elbow Method')
plt.xlabel('Number of Clusters')
plt.ylabel('WCSS')
plt.show()
```

```python
# Step 4 - K-Means Clustering

from sklearn.cluster import KMeans

kmeans_model = KMeans(
    n_clusters=3,
    random_state=0,
    n_init=10
)

data['Cluster'] = kmeans_model.fit_predict(X_cluster_scaled)

# Number of students in each cluster
print(data['Cluster'].value_counts().sort_index())

# Average scores for each cluster
print(data.groupby('Cluster')[['math score',
                               'reading score',
                               'writing score']].mean().round(2))
```

```python
# Step 5 - Visualize the Clusters

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=data,
    x='math score',
    y='reading score',
    hue='Cluster',
    palette='Set1'
)

plt.title('K-Means Clustering of Student Performance')
plt.xlabel('Math Score')
plt.ylabel('Reading Score')
plt.show()
```

```python
# Cluster Summary

print("Number of students in each cluster:")
print(data['Cluster'].value_counts().sort_index())

print("\nAverage scores for each cluster:")
print(
    data.groupby('Cluster')[
        ['math score', 'reading score', 'writing score']
    ].mean().round(2)
)
```
